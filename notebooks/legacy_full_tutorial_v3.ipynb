## 3 · Forward Projection


X-ray: polychromatic Beer-Lambert sum (beam hardening emerges naturally).
Neutron: three-component thermal model (absorption + coherent + incoherent scatter).


````python
KVP       = 120.0
AL_FILTER = 2.0

xray_sino, neut_sino = nxs.make_sinogram_pair(
    phantom,
    n_angles        = N_ANGLES,
    angle_range_deg = 360.0,
    kVp             = KVP,
    filter_mm_Al    = AL_FILTER,
    n_spectrum_bins = 12,
    I0_xray         = 1e5,
    I0_neutron      = 1e5,
    use_astra       = True,
)
print('X-ray sinogram :', xray_sino['sino_lam'].shape,
      '  (n_angles, N_slice, N_det)')
print('Neutron sinogram:', neut_sino['sino_lam'].shape)
````

````python
# ── Visualise central sinogram slices ─────────────────────────────────────
s = N // 2
fig, axes = plt.subplots(1, 2, figsize=(11, 4))
for ax, sino, title in zip(axes,
        [xray_sino['sino_lam'][:, s, :], neut_sino['sino_lam'][:, s, :]],
        ['X-ray log-attenuation sinogram', 'Neutron log-attenuation sinogram']):
    im = ax.imshow(sino, aspect='auto', cmap='gray', origin='lower')
    ax.set_xlabel('Detector pixel'); ax.set_ylabel('Projection angle')
    ax.set_title(title)
    plt.colorbar(im, ax=ax, label='OD')
plt.tight_layout(); plt.show()
````

## 5 · CT Reconstruction


Nine algorithms are available. Select one with `ALGORITHM` below.

| Algorithm | Type | GPU | Notes |
|-----------|------|-----|-------|
| `FBP` | analytic | optional | Fast; ramp filter; ringing artefacts |
| `GRIDREC` | analytic | no | Fourier gridding; needs TomoPy |
| `SIRT` | iterative | yes | Best H₂O/HDPE separation |
| `SART` | iterative | yes | Recommended default |
| `CGLS` | iterative | yes | Fast; needs regularisation |
| `EM` | iterative | yes | ML-EM; enforces positivity |
| `OSSART` | iterative | yes | Ordered-subset SART; fastest iterative |
| `TV_MIN` | iterative | yes | Total-variation; piecewise-constant |
| `NESTEROV_SIRT` | iterative | yes | Nesterov momentum; ~1.7× faster than SIRT |

```python
from neutron_xray_sim import AVAILABLE_ALGORITHMS
print(AVAILABLE_ALGORITHMS)
```


````python
from neutron_xray_sim.artifacts import inject_sinogram_artifacts, inject_volume_artifacts

# ── Inject sinogram-domain artifacts ──────────────────────────────────────
x_sino_art, n_sino_art = inject_sinogram_artifacts(
    {k: v.copy() if isinstance(v, np.ndarray) else v for k, v in xray_sino.items()},
    {k: v.copy() if isinstance(v, np.ndarray) else v for k, v in neut_sino.items()},
    custom_cfg,
)
````

````python
# ── Choose algorithm and reconstruct ──────────────────────────────────────
ALGORITHM   = 'SART'         # change to any of the nine algorithms above
FILTER_NAME = 'shepp-logan'  # for FBP / GRIDREC only
N_ITER      = 30             # for all iterative algorithms
N_SUBSETS   = 10             # for OSSART only
LAMBDA_TV   = 0.02           # for TV_MIN only

vol_x, vol_n = nxs.reconstruct_pair(
    x_sino_art, n_sino_art,
    algorithm    = ALGORITHM,
    filter_name  = FILTER_NAME,
    n_iter       = N_ITER,
    n_subsets    = N_SUBSETS,
    lambda_tv    = LAMBDA_TV,
    remove_rings = True,
    use_astra    = True,
    clip_negative= True,
)

# Apply volume-domain artifacts (misalignment etc.)
vol_x, vol_n = inject_volume_artifacts(vol_x, vol_n, custom_cfg)

print(f'vol_x: {vol_x.shape}  [{vol_x.min():.3f}, {vol_x.max():.3f}] cm⁻¹')
print(f'vol_n: {vol_n.shape}  [{vol_n.min():.3f}, {vol_n.max():.3f}] cm⁻¹')
````

````python
# ── Reconstructed central slices ──────────────────────────────────────────
si = N // 2
fig, axes = plt.subplots(2, 3, figsize=(11, 7))
for col, (vx_sl, vn_sl, lbl) in enumerate([
        (vol_x[si],       vol_n[si],       'XY'),
        (vol_x[:, si, :], vol_n[:, si, :], 'XZ'),
        (vol_x[:, :, si], vol_n[:, :, si], 'YZ'),
]):
    vmin_x, vmax_x = np.percentile(vol_x, [1, 99])
    vmin_n, vmax_n = np.percentile(vol_n, [1, 99])
    axes[0, col].imshow(vx_sl, cmap='gray', vmin=vmin_x, vmax=vmax_x)
    axes[0, col].set_title(f'X-ray {lbl} — {ALGORITHM}'); axes[0, col].axis('off')
    axes[1, col].imshow(vn_sl, cmap='gray', vmin=vmin_n, vmax=vmax_n)
    axes[1, col].set_title(f'Neutron {lbl} — {ALGORITHM}'); axes[1, col].axis('off')
plt.tight_layout(); plt.show()
````

````python
DISPLACEMENTS = [0.0, 0.5, 1.0, 2.0, 3.0, 5.0, 8.0]   # voxels

sim_sweep = nxs.DualModalitySimulation(
    preset=PRESET, N=N, n_angles=N_ANGLES,
    algorithm='SART', verbose=False,
)

sweep_results = []
ref_r = sim_sweep.run(nxs.ArtifactConfig.clean(), tag='0 vx (clean)')
sweep_results.append(ref_r)

for d in DISPLACEMENTS[1:]:
    cfg_d = nxs.ArtifactConfig(
        misalignment=True,
        translation_voxels=(d, 0.0, 0.0),
    )
    r = sim_sweep.run(cfg_d, tag=f'{d:.1f} vx', ref_result=ref_r)
    sweep_results.append(r)
````

````python
# ── Streak score vs displacement ──────────────────────────────────────────
scores = [r.signatures.horizontal_streak_score for r in sweep_results]

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4.5))
ax1.plot(DISPLACEMENTS, scores, 'o-', color='tomato', lw=2, ms=7)
ax1.set_xlabel('Translation misalignment [voxels]')
ax1.set_ylabel('Horizontal streak score')
ax1.set_title('Streak score vs misalignment')
ax1.grid(True, alpha=0.3)

# Overlay the 0-vx and 3-vx histograms for comparison
for r, col, lbl in [
        (sweep_results[0], 'steelblue', '0 vx'),
        (sweep_results[DISPLACEMENTS.index(3.0)], 'tomato', '3 vx'),
]:
    H = r.histogram.H.T
    H = np.log1p(np.ma.masked_where(H == 0, H))
    ax2.contour(H, levels=6, colors=[col], alpha=0.7,
                extent=r.histogram.extent)
    ax2.plot([], [], color=col, label=lbl)
ax2.set_xlabel(r'$\mu_x$ [cm$^{-1}$]'); ax2.set_ylabel(r'$\mu_n$ [cm$^{-1}$]')
ax2.set_title('Histogram contours: 0 vx vs 3 vx')
ax2.legend(); ax2.grid(True, alpha=0.3)
plt.tight_layout(); plt.show()
````

## 10 · File-System Cache — Save and Reload Every Pipeline Stage  ★ new in v3


The `SimCache` class saves every pipeline stage as `.npy` files.
On the next run, expensive steps (phantom build, GPU projection,
iterative reconstruction) are skipped if the files already exist.

**Directory layout:**
```
results/composite_SART/
├── phantom/               label_vol.npy, mu_x_vol_*.npy, mu_n_*_vol.npy, meta.json
├── sinograms/             xray_sino_lam.npy, neutron_sino_lam.npy, sino_meta.json
└── runs/
    └── clean/             vol_xray.npy, vol_neutron.npy,
                           histogram_H.npy, histogram_*edges.npy, run_meta.json
```

After the survey is saved you can reload any stage independently
and re-run only the analysis — no GPU required.


````python
CACHE_DIR = 'results/composite_SART_cache'

# ── Run with cache enabled ─────────────────────────────────────────────────
sim_cached = nxs.DualModalitySimulation(
    preset    = PRESET,
    N         = N,
    n_angles  = N_ANGLES,
    algorithm = 'SART',
    verbose   = True,
    cache_dir       = CACHE_DIR,  # ← enable cache
    overwrite_cache = True,       # overwrite if already exists
)

rc_clean = sim_cached.run(nxs.ArtifactConfig.clean(),           tag='clean')
rc_noise = sim_cached.run(nxs.ArtifactConfig.noise_only(I0=5e4), tag='noise_moderate')
rc_real  = sim_cached.run(nxs.ArtifactConfig.realistic(),        tag='realistic')

print('\nSaved tags:', sim_cached.cache.list_run_tags())
````

````python
# ── Inspect the cache directory ────────────────────────────────────────────
import os
for root, dirs, files in os.walk(CACHE_DIR):
    level = root.replace(CACHE_DIR, '').count(os.sep)
    indent = '  ' * level
    print(f'{indent}{os.path.basename(root)}/')
    for f in sorted(files):
        size = os.path.getsize(os.path.join(root, f)) / 1e6
        print(f'{indent}  {f:45s}  {size:.2f} MB')
````

| Metric | Symbol | Meaning | Better |

| Cluster spread | σ_x, σ_n | Std of cluster along each axis [cm⁻¹] | lower |
| Davies-Bouldin | DB | Separability index (compactness ÷ separation) | lower |

**`evaluate_histogram_quality(hist, phantom)`** returns all three.
**`compare_algorithms(results, phantom)`** tabulates them across a list

````python
 
import gc
import json
import time
import ctypes
import platform
import warnings
from pathlib import Path
 
import numpy as np
 
try:
    import psutil as _psutil
    _PSUTIL = True
except ImportError:
    _PSUTIL = False
    warnings.warn(
        "psutil not found — RAM monitoring disabled.  "
        "Install with:  pip install psutil"
    )
 
import neutron_xray_sim as nxs
from neutron_xray_sim.artifacts import (
    inject_sinogram_artifacts,
    inject_volume_artifacts,
)
 
 
# ─────────────────────────────────────────────────────────────────────────────
# Memory utilities
# ─────────────────────────────────────────────────────────────────────────────
 
def _ram_mb():
    if not _PSUTIL:
        return 0.0
    return _psutil.Process().memory_info().rss / 1e6
 
def _system_ram_pct():
    if not _PSUTIL:
        return 0.0
    return _psutil.virtual_memory().percent
 
def _system_ram_available_gb():
    if not _PSUTIL:
        return float("nan")
    return _psutil.virtual_memory().available / 1e9
 
def _free_ram():
    """
    Force Python to release memory back to the OS.
 
    Runs three GC generations, then calls:
    - Linux : malloc_trim(0) — returns freed glibc heap pages to the kernel
              immediately, rather than waiting for the next large allocation.
    - Windows: HeapCompact (best-effort; Windows already returns most memory
              promptly when large allocations are freed).
 
    Call this after every explicit del of a large array.
    """
    gc.collect(0)
    gc.collect(1)
    gc.collect(2)
    if platform.system() == "Linux":
        try:
            ctypes.CDLL("libc.so.6").malloc_trim(0)
        except Exception:
            pass
    elif platform.system() == "Windows":
        try:
            ctypes.windll.kernel32.HeapCompact(0, 0)
        except Exception:
            pass
 
def _astra_clear():
    """
    Release all ASTRA GPU objects (data arrays, algorithms, projectors).
 
    ASTRA keeps a global registry; if a previous run crashed mid-slice,
    orphaned GPU buffers remain allocated.  Calling this before each
    reconstruction ensures a clean GPU state and prevents VRAM leaks.
    """
    try:
        import astra
        astra.data2d.clear()
        astra.data3d.clear()
        astra.algorithm.clear()
        astra.projector.clear()
    except Exception:
        pass
 
 
# ─────────────────────────────────────────────────────────────────────────────
# Main batch function
# ─────────────────────────────────────────────────────────────────────────────
 
def batch_reconstruct_all(
    output_dir       = "batch_1024",
    phantoms         = None,
    algorithms       = None,
    artifacts        = None,
    N                = 1024,
    n_angles         = 1200,
    kVp              = 120.0,
    filter_mm_Al     = 2.0,
    n_iter           = 30,
    n_subsets        = 10,
    lambda_tv        = 0.02,
    filter_name      = "shepp-logan",
    remove_rings     = True,
    ram_warn_gb      = 300.0,
    ram_abort_gb     = None,
    use_astra        = True,
    overwrite        = False,
    rng_seed         = 0,
):
    """
    Run every combination of phantom x artifact x algorithm for X-ray
    and neutron, saving reconstructed volumes as .npy files.
 
    Memory design
    -------------
    - Sinogram computed ONCE per (phantom, artifact), reused for all algorithms.
    - Raw sinogram copies deleted immediately after artifact injection.
    - Both modalities reconstructed together then inject_volume_artifacts called
      once with REAL volumes (no dummy arrays).
    - Explicit del + _free_ram() after every large temporary array.
    - ASTRA GPU objects cleared before every reconstruction call.
    - RAM monitored after every algorithm step via psutil.
 
    Parameters
    ----------
    output_dir    : root output directory
    phantoms      : list of preset names, or None = all four
    algorithms    : list of algorithm names, or None = all nine
    artifacts     : list of (label, kwargs) or None = default 11
    N             : voxel grid N x N x N
    n_angles      : projection angles
    kVp           : X-ray tube voltage [kV]
    filter_mm_Al  : Al pre-filter [mm]
    n_iter        : iterations for iterative algorithms
    n_subsets     : ordered subsets for OSSART
    lambda_tv     : TV weight for TV_MIN
    filter_name   : ramp filter for FBP/GRIDREC
    remove_rings  : Vo ring removal pre-processing
    ram_warn_gb   : warn when process RAM exceeds this [GB]
    ram_abort_gb  : raise MemoryError if RAM exceeds this [GB] (None=off)
    use_astra     : use ASTRA GPU
    overwrite     : skip existing volumes if False
    rng_seed      : reproducibility seed
 
    Returns
    -------
    manifest : list of dicts with keys
        phantom, artifact, algorithm, modality,
        path, shape, ram_mb_after, elapsed_s, skipped
    """
 
    # ── Defaults ──────────────────────────────────────────────────────────────
    if phantoms is None:
        phantoms = ["composite", "battery", "bone_implant", "industrial"]
 
    if algorithms is None:
        algorithms = list(nxs.AVAILABLE_ALGORITHMS)
 
    if artifacts is None:
        artifacts = [
            ("clean",
             {}),
            ("noise_moderate",
             dict(photon_noise=True, I0_xray=5e4, I0_neutron=5e4)),
            ("noise_low_dose",
             dict(photon_noise=True, I0_xray=5e3, I0_neutron=5e3)),
            ("beam_hardening",
             dict(apply_bh_correction=False)),
            ("bh_corrected",
             dict(apply_bh_correction=True, bh_correction_order=3)),
            ("neutron_scatter",
             dict(neutron_scatter=True, scatter_fraction=0.06,
                  scatter_sigma_pixels=9.0)),
            ("xray_scatter",
             dict(xray_scatter=True, xray_scatter_fraction=0.04)),
            ("detector_psf",
             dict(detector_psf=True,
                  psf_sigma_xray_pixels=0.8,
                  psf_sigma_neutron_pixels=1.5)),
            ("ring_artifacts",
             dict(ring_artifacts=True, n_bad_columns=3,
                  ring_amplitude=0.05)),
            ("misalignment",
             dict(misalignment=True,
                  translation_voxels=(3.0, 0.0, 0.0),
                  rotation_deg=(0.0, 1.0, 0.0))),
            ("all_realistic",
             dict(photon_noise=True,    I0_xray=5e4,   I0_neutron=5e4,
                  neutron_scatter=True, scatter_fraction=0.06,
                  xray_scatter=True,    xray_scatter_fraction=0.04,
                  detector_psf=True,
                  psf_sigma_xray_pixels=0.8,
                  psf_sigma_neutron_pixels=1.5,
                  misalignment=True,
                  translation_voxels=(3.0, 0.0, 0.0))),
        ]
 
    root = Path(output_dir)
    root.mkdir(parents=True, exist_ok=True)
    rng  = np.random.default_rng(rng_seed)
 
    total_pairs  = len(phantoms) * len(artifacts) * len(algorithms)
    total_vols   = total_pairs * 2
    done_pairs   = 0
    manifest     = []
    t_global     = time.perf_counter()
    vol_mb       = (N ** 3 * 4) / 1e6
 
    # ── Header ────────────────────────────────────────────────────────────────
    print("=" * 76)
    print("  Memory-efficient batch reconstruction")
    print(f"  Phantoms   : {phantoms}")
    print(f"  Algorithms : {algorithms}")
    print(f"  Artifacts  : {[a[0] for a in artifacts]}")
    print(f"  N={N}  angles={n_angles}  iter={n_iter}  overwrite={overwrite}")
    print(f"  Volume     : {vol_mb:.1f} MB each  |  "
          f"Max simultaneous: 2 volumes = {2*vol_mb:.0f} MB")
    print(f"  Total      : {total_vols} volumes  "
          f"({total_vols*vol_mb/1e3:.1f} GB if kept — we keep 2)")
    
    send_message("  Memory-efficient batch reconstruction")
    send_message(f"  Phantoms   : {phantoms}")
    send_message(f"  Algorithms : {algorithms}")
    send_message(f"  Artifacts  : {[a[0] for a in artifacts]}")
    send_message(f"  N={N}  angles={n_angles}  iter={n_iter}  overwrite={overwrite}")
    send_message(f"  Volume     : {vol_mb:.1f} MB each  |  "
          f"Max simultaneous: 2 volumes = {2*vol_mb:.0f} MB")
    send_message(f"  Total      : {total_vols} volumes  "
          f"({total_vols*vol_mb/1e3:.1f} GB if kept — we keep 2)")
    if _PSUTIL:
        vm = _psutil.virtual_memory()
        print(f"  System RAM : {vm.total/1e9:.1f} GB total  |  "
              f"{vm.available/1e9:.1f} GB available now")
        print(f"  RAM warn={ram_warn_gb} GB"
              + (f"  abort={ram_abort_gb} GB" if ram_abort_gb else "  abort=off"))
        send_message(f"  System RAM : {vm.total/1e9:.1f} GB total  |  "
              f"{vm.available/1e9:.1f} GB available now")
        send_message(f"  RAM warn={ram_warn_gb} GB"
              + (f"  abort={ram_abort_gb} GB" if ram_abort_gb else "  abort=off"))
    else:
        print("  RAM monitoring: disabled (pip install psutil)")
    print("=" * 76)
 
    def _check_ram(label):
        rss_mb   = _ram_mb()
        rss_gb   = rss_mb / 1e3
        avail_gb = _system_ram_available_gb()
        sys_pct  = _system_ram_pct()
        if _PSUTIL:
            line = (f"  RAM  proc={rss_gb:.2f} GB  "
                    f"sys={sys_pct:.0f}%  avail={avail_gb:.1f} GB  [{label}]")
            if rss_gb > ram_warn_gb:
                print("  WARNING " + line)
                send_message("  WARNING " + line)
            else:
                print("        " + line)
                send_message("        " + line)
            if ram_abort_gb and rss_gb > ram_abort_gb:
                raise MemoryError(
                    f"Process RAM {rss_gb:.2f} GB > abort threshold "
                    f"{ram_abort_gb} GB at [{label}]."
                )
                send_message(f"Process RAM {rss_gb:.2f} GB > abort threshold "
                    f"{ram_abort_gb} GB at [{label}].")
        return rss_mb
 
    # ═════════════════════════════════════════════════════════════════════════
    # Phantom loop
    # ═════════════════════════════════════════════════════════════════════════
    for ph_name in phantoms:
 
        print(f"\n{'─'*76}")
        print(f"  [Phantom] {ph_name}  N={N}")
        send_message(f"  [Phantom] {ph_name}  N={N}")
        _free_ram()
        phantom = nxs.make_phantom(ph_name, N)
        (root / ph_name).mkdir(exist_ok=True)
 
        # ── Artifact loop ─────────────────────────────────────────────────────
        for art_label, art_kwargs in artifacts:
 
            art_dir = root / ph_name / art_label
            art_dir.mkdir(exist_ok=True)
            cfg = nxs.ArtifactConfig(**art_kwargs)
 
            sino_x_path = art_dir / "raw_xray_sino.npy"
            sino_n_path = art_dir / "raw_neutron_sino.npy"
            ang_path    = art_dir / "angles_deg.npy"
            vcm_path    = art_dir / "voxel_cm.txt"
 
            if sino_x_path.exists() and not overwrite:
                print(f"  [sino] {ph_name}/{art_label}  loading from disk")
                x_sino_lam = np.load(str(sino_x_path))
                n_sino_lam = np.load(str(sino_n_path))
                angles_deg = np.load(str(ang_path))
                voxel_cm   = float(vcm_path.read_text().strip())
            else:
                print(f"  [sino] {ph_name}/{art_label}  projecting...",
                      end=" ", flush=True)
                t0 = time.perf_counter()
 
                xray_raw, neut_raw = nxs.make_sinogram_pair(
                    phantom, n_angles=n_angles, kVp=kVp,
                    filter_mm_Al=filter_mm_Al, use_astra=use_astra,
                )
 
                x_art, n_art = inject_sinogram_artifacts(
                    {k: v.copy() if isinstance(v, np.ndarray) else v
                     for k, v in xray_raw.items()},
                    {k: v.copy() if isinstance(v, np.ndarray) else v
                     for k, v in neut_raw.items()},
                    cfg, rng=rng,
                )
 
                # Free raw sinograms — largest arrays, no longer needed
                del xray_raw, neut_raw
                _free_ram()
 
                x_sino_lam = x_art["sino_lam"]
                n_sino_lam = n_art["sino_lam"]
                angles_deg = x_art["angles_deg"]
                voxel_cm   = float(x_art.get("voxel_cm", phantom.voxel_cm))
 
                np.save(str(sino_x_path), x_sino_lam)
                np.save(str(sino_n_path), n_sino_lam)
                np.save(str(ang_path),    angles_deg)
                vcm_path.write_text(str(voxel_cm))
 
                # Free dict wrappers (arrays kept via x_sino_lam / n_sino_lam)
                del x_art, n_art
                _free_ram()
 
                print(f"{time.perf_counter()-t0:.1f}s")
                send_message(f"{time.perf_counter()-t0:.1f}s")
 
            x_sino_dict = {"sino_lam": x_sino_lam,
                           "angles_deg": angles_deg,
                           "voxel_cm": voxel_cm}
            n_sino_dict = {"sino_lam": n_sino_lam,
                           "angles_deg": angles_deg,
                           "voxel_cm": voxel_cm}
 
            # ── Algorithm loop ────────────────────────────────────────────────
            for alg in algorithms:
 
                done_pairs += 1
                pct    = 100.0 * done_pairs / total_pairs
                t_tot  = time.perf_counter() - t_global
                eta_s  = (t_tot / done_pairs) * (total_pairs - done_pairs)
 
                alg_dir = art_dir / alg
                alg_dir.mkdir(exist_ok=True)
                x_path  = alg_dir / "vol_xray.npy"
                n_path  = alg_dir / "vol_neutron.npy"
 
                # Skip if both exist
                if x_path.exists() and n_path.exists() and not overwrite:
                    print(f"  [{pct:5.1f}%]  SKIP  "
                          f"{ph_name}/{art_label}/{alg}  "
                          f"ETA {eta_s/60:.1f} min")
                    rss = _ram_mb()
                    for mod, p in [("xray", x_path), ("neutron", n_path)]:
                        manifest.append({
                            "phantom": ph_name, "artifact": art_label,
                            "algorithm": alg,   "modality": mod,
                            "path": str(p),     "shape": None,
                            "ram_mb_after": rss, "elapsed_s": 0.0,
                            "skipped": True,
                        })
                    continue
 
                t_alg = time.perf_counter()
 
                # Clear any stale ASTRA GPU objects before reconstruction
                _astra_clear()
                vol_x = nxs.reconstruct(
                    x_sino_dict, algorithm=alg, filter_name=filter_name,
                    n_iter=n_iter, n_subsets=n_subsets, lambda_tv=lambda_tv,
                    remove_rings=remove_rings, use_astra=use_astra,
                    clip_negative=True,
                )
 
                _astra_clear()
                vol_n = nxs.reconstruct(
                    n_sino_dict, algorithm=alg, filter_name=filter_name,
                    n_iter=n_iter, n_subsets=n_subsets, lambda_tv=lambda_tv,
                    remove_rings=remove_rings, use_astra=use_astra,
                    clip_negative=True,
                )
                _astra_clear()
 
                # Volume-domain artifacts — one call with BOTH real volumes
                vol_x, vol_n = inject_volume_artifacts(vol_x, vol_n, cfg, rng=rng)
 
                np.save(str(x_path), vol_x)
                np.save(str(n_path), vol_n)
 
                elapsed = time.perf_counter() - t_alg
 
                # Free volumes before moving to next algorithm
                del vol_x, vol_n
                _free_ram()
 
                # RAM check + progress line
                rss_mb = _check_ram(f"{ph_name}/{art_label}/{alg}")
                t_tot  = time.perf_counter() - t_global
                eta_s  = (t_tot / done_pairs) * (total_pairs - done_pairs)
                send_message(str(done_pairs)+" Done")
                send_message(str(total_pairs - done_pairs)+" Remaining")
                print(f"  [{pct:5.1f}%]  "
                      f"{ph_name}/{art_label}/{alg}  "
                      f"{elapsed:.1f}s  ETA {eta_s/60:.1f} min")
                send_message(f"  [{pct:5.1f}%]  "
                      f"{ph_name}/{art_label}/{alg}  "
                      f"{elapsed:.1f}s  ETA {eta_s/60:.1f} min")
 
                for mod, p in [("xray", x_path), ("neutron", n_path)]:
                    manifest.append({
                        "phantom":      ph_name,
                        "artifact":     art_label,
                        "algorithm":    alg,
                        "modality":     mod,
                        "path":         str(p),
                        "shape":        (N, N, N),
                        "ram_mb_after": rss_mb,
                        "elapsed_s":    elapsed / 2,
                        "skipped":      False,
                    })
 
            # After all algorithms: free sinogram arrays
            del x_sino_lam, n_sino_lam, x_sino_dict, n_sino_dict
            _free_ram()
 
        # After all artifacts: free phantom
        del phantom
        _free_ram()
 
    # ── Summary ───────────────────────────────────────────────────────────────
    elapsed_total = time.perf_counter() - t_global
    n_done    = sum(1 for m in manifest if not m["skipped"])
    n_skipped = sum(1 for m in manifest if m["skipped"])
 
    print("\n" + "=" * 76)
    print("  Batch complete.")
    print(f"  Reconstructed : {n_done}")
    print(f"  Skipped       : {n_skipped}")
    print(f"  Total time    : {elapsed_total/60:.1f} min")
    if _PSUTIL:
        print(f"  Final RAM     : proc={_ram_mb()/1e3:.2f} GB  "
              f"avail={_system_ram_available_gb():.1f} GB")
    print(f"  Output dir    : {root.resolve()}")
    print("=" * 76)

    send_message("  Batch complete.")
    send_message(f"  Reconstructed : {n_done}")
    send_message(f"  Skipped       : {n_skipped}")
    send_message(f"  Total time    : {elapsed_total/60:.1f} min")
    if _PSUTIL:
        send_message(f"  Final RAM     : proc={_ram_mb()/1e3:.2f} GB  "
              f"avail={_system_ram_available_gb():.1f} GB")
    send_message(f"  Output dir    : {root.resolve()}")
 
    manifest_path = root / "manifest.json"
    with open(manifest_path, "w") as f:
        json.dump(manifest, f, indent=2, default=str)
    print(f"  Manifest: {manifest_path}")
    send_message(f"  Manifest saved: {manifest_path}")
    
    return manifest
````
